```python
import zipcodes
```

```python
    
def get_zip_codes(city_name, state_abbr):
    """
    Retrieves a list of zip codes for a given city and state.

    Args:
        city_name (str): The name of the city.
        state_abbr (str): The two-letter abbreviation of the state.

    Returns:
        list: A list of zip codes for the city and state, or an empty list if not found.
    """
    zip_codes = zipcodes.filter_by(city=city_name, state=state_abbr)
    return [z['zip_code'] for z in zip_codes]

```
