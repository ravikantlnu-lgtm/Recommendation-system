```python
import pandas as pd
from google.cloud import bigquery
import pickle
```

```python
# Project ID where the BigQuery table resides
project_a = "proj-docai-dev"  # Replace with your Project A ID
dataset_id = "sales_rec_demo"  # Replace with your dataset ID
table_id = "cc_data_curation"      # Replace with your table ID
pickle_file_path = "data/construct_connect_data.pkl"  # Path to save the pickled DataFrame
```

```python

# Construct a BigQuery client object.
client = bigquery.Client(project=project_a)

query = f"SELECT * FROM `{project_a}.{dataset_id}.{table_id}` limit 10000"
query_job = client.query(query)

# Convert the query result to a Pandas DataFrame
df = query_job.to_dataframe()

```

```python
df
```

```python
df.to_csv('data/cc_data_curation.csv')
```

```python

# Pickle the DataFrame
with open(pickle_file_path, "wb") as f:
    pickle.dump(df, f)
```
