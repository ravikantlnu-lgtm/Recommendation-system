```python
import pandas as pd
```

```python
df_single_prompt_output = pd.read_csv('data/single_prompt_output.csv')
df_single_prompt_output
```

```python
df_multi_prompt_output = pd.read_csv('data/second_pass_output.csv')
df_multi_prompt_output
```

```python
merged_df = pd.merge(df_single_prompt_output, df_multi_prompt_output, on=['ProjectID', 'Product'], suffixes=('_single', '_multi'), how='outer')
merged_df
```

```python
merged_df['match'] = merged_df['Relevance Classification_single'] == merged_df['Relevance Classification_multi']
merged_df
```

```python
merged_df['match'] = merged_df['match'].fillna(False)  # Treat NaN as not matching
merged_df
```

```python
merged_df['match'].sum()
```

```python
merged_df.to_csv('data/merged_output.csv', index=False)
```
