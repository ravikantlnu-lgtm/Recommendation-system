````python
import pandas as pd
import json
````

````python
def read_and_save_boolean_filters():
    boolean_df = pd.DataFrame(columns=['Filter', 'Query'])
    with open('data/boolean_filters_3.txt', 'r') as f:
            boolean_filters = f.read()
    b_filters = json.loads(boolean_filters.replace('```', '').replace('json', '').replace("{'", '{"').replace("':", '":').replace(" '", ' "').replace('", "', '"}, {"'))
    f_dict = {k: v for d in b_filters for k, v in d.items()}
    keys = f_dict.keys()
    values = f_dict.values()
    boolean_df['Filter'] = keys
    boolean_df['Query'] = values

    df_ordered = boolean_df.sort_values(by='Filter')
    df_ordered.to_csv('data/boolean_filters.csv', index = False)
    #save f_dict to json file
    with open('data/boolean_filters.json', 'w') as f:
        json.dump(f_dict, f)
````

````python
def get_boolean_filter_df():
    boolean_df = pd.read_csv('data/boolean_filters_latest.csv')
    return boolean_df

def get_boolean_filter_json():
    
    with open('data/boolean_filters_latest.json', 'r', encoding='utf-8') as f:  # Use 'r' for reading
            data = json.dumps(f)  ## Load the JSON data
            print(data.type)
    return data
````

````python
def get_stage_filter_json():
    stage_filter = {"high": ["Biddate Set", "Construction Documents", "General Contractor Award", "Low Bids Announced", "SUBBIDS: ASAP"], "moderate": ["Construction Underway", "Post Bid"], "ignore": ["Cancelled", "Design Development", "Disqualified Lead", "Duplicate Project", "Pre-Design", "Schematic Design"]}
    return stage_filter
````

````python
def get_relevant_materials():
    materials = pd.read_csv('data/relevant_materials.csv')
    materials_json = materials.to_json(orient='records')
    return materials_json
````

````python
def get_all_filters():
    stage_filter = get_stage_filter_json()
    materials = get_relevant_materials()
    return stage_filter, materials
````
